```python
import pandas as pd
import numpy as np
df=pd.DataFrame(np.random.randn(4,3),index=['a','c','d','f'],columns=['one','two','three'])
df=df.reindex(['a','b','c','d','e','f'])
print("original Dataframe with NaN\n",df)
print("\n")
print("dropped DataFrame\n",df.dropna())
print("\n")
```
