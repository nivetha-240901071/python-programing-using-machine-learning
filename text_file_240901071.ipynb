```python
f=open("textfile.txt",'w')
f.write("first line\n")
f.write("second line\n")
f.write("third line\n")
f.close()
k=open("textfile.txt",'r')
print(k.readline())
print(k.readline())
print(k.readline())
k.close()
```

```python
m=open("textfile.txt",'a')
m.write("fourth line\n")
m.write("fifth line\n")
m.close()

p=open("textfile.txt",'r')
print(p.read())
p.close()
```
